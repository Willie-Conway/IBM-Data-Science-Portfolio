# Import Libararies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

# Data Handling

```python
from sklearn.datasets import fetch_california_housing
```

```python
df = fetch_california_housing()
df
```

```python
california_housing = pd.DataFrame(df['data'],columns=['MedInc','HouseAge','AveRooms','AveBedrms','Population','AveOccup','Latitude','Longitude'])
california_housing.head()
```

```python
california_housing['MedHouseVal']= df["target"]
```

```python
california_housing.head()
```

```python
california_housing.info()
```

```python
california_housing.isna().sum()
```

```python
x = california_housing.drop("MedHouseVal",axis=1)
x.head()
```

```python
y=california_housing["MedHouseVal"]
y.head()
```

# Trying Differrent Moudeles

```python
from sklearn.model_selection import train_test_split
x_train , x_test ,y_train ,y_test =train_test_split(x,y,test_size=0.2)
from sklearn.linear_model import Ridge
module_1 =Ridge()
module_1.fit(x_train,y_train)
module_1.score(x_test,y_test)
```

```python
from sklearn.linear_model import Lasso
module_3=Lasso()
module_3.fit(x_train,y_train)
module_3.score(x_test,y_test)
```

```python
from sklearn.linear_model import ElasticNet
module_4=ElasticNet()
module_4.fit(x_train,y_train)
module_4.score(x_test,y_test)
```

```python
from sklearn.ensemble import RandomForestRegressor
module_5=RandomForestRegressor()
module_5.fit(x_train,y_train)
module_5.score(x_test,y_test)
```

```python
from sklearn.neighbors import KNeighborsRegressor
model_6 = KNeighborsRegressor(weights='distance',
 p = 1,
 n_neighbors = 8,
 leaf_size = 15,
 algorithm = 'brute')
model_6.fit(x_train,y_train)
model_6.score(x_test,y_test)
```

```python
from sklearn.model_selection import RandomizedSearchCV
grid = {"n_neighbors": [1,2,3,4,5,6,8,9,10],
        "weights": ['uniform', 'distance'],
        "p": [1, 2],
        "algorithm":  ['auto', 'ball_tree', 'kd_tree', 'brute'],
        "leaf_size": [15, 20,30]}
np.random.seed(42)
from sklearn.model_selection import train_test_split
X_train ,X_test ,Y_train ,Y_test = train_test_split(x,y,test_size=0.2,shuffle=True)
from sklearn.neighbors import KNeighborsRegressor
model =KNeighborsRegressor(n_jobs=-1)
Search_cv =RandomizedSearchCV(estimator=model
                             ,param_distributions=grid
                             ,n_iter=20
                             ,cv=5
                             ,verbose=2)
Search_cv.fit(x_train,y_train)
```

```python
Search_cv.best_params_
```
